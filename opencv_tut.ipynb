```python
#Cropping
img5 = cv2.imread(r"E:\New folder\OneDrive\Desktop\figuresssweek44444\fig_01.png")
img5_resi = cv2.resize(img5,(512,512))

flag = False
ix=-1
iy=-1

def crop(event,x,y,flags,params):
    global flag,ix,iy

    if event == cv2.EVENT_LBUTTONDOWN:
        flag = True
        ix=x
        iy=y

    elif event == cv2.EVENT_LBUTTONUP:
        fx = x
        fy = y

        flag = False
        cv2.rectangle(img5_resi,pt1=(ix,iy),pt2=(x,y),thickness=1,color=(0,255,0))
        #crop tools
        cropped = img5_resi[iy:fy,ix:fx]
        cv2.imshow("new_window",cropped)
        cv2.waitKey(0)
        

cv2.namedWindow(winname = "window")
cv2.setMouseCallback("window",crop)


while True:
    cv2.imshow('window',img5_resi)
    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

cv2.destroyAllWindows()
```
